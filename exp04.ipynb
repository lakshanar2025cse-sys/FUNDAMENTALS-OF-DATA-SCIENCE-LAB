```python
import numpy as np
import pandas as pd

data = {
    "Department": [
        "CSE", "ECE", "IT", "CSE", "ECE",
        "IT", "CSE", np.nan, "ECE", "IT"
    ],

    "Experience": [
        2, 5, 3, np.nan, 7,
        4, 1, 6, 3, 5
    ],

    "Salary": [
        30000, 45000, np.nan, 38000, 60000,
        50000, 28000, 65000, 42000, 55000
    ],

    "Promoted": [
        "Yes", "No", "Yes", "No", "Yes",
        "No", "Yes", "Yes", "No", "Yes"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("\nDataset Information:")
df.info()

print("\nMode of Department:")
print(df["Department"].mode())

df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)

df["Experience"] = df["Experience"].fillna(
    df["Experience"].median()
)
 
df["Salary"] = df["Salary"].fillna(
    round(df["Salary"].mean())
)

print("\nAfter Data Preprocessing:")
print(df)
```
